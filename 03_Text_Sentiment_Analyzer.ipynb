```python
# ============================================================
# APPLICATION 3
# TEXT SENTIMENT ANALYZER
# Text & Speech Analysis
# Google Colab
# ============================================================

# Install required library
!pip -q install textblob

from textblob import TextBlob
from IPython.display import display, HTML


# ============================================================
# APPLICATION HEADER
# ============================================================

display(HTML("""
<div style="
background:#f5f7fb;
padding:30px;
border-radius:20px;
text-align:center;
font-family:Arial;
">

<h1>Text Sentiment Analyzer</h1>

<p style="font-size:18px;">
Analyze the emotional tone of text
</p>

<p>
Text & Speech Analysis - Application 3
</p>

</div>
"""))


# ============================================================
# GET USER INPUT
# ============================================================

print("\nEnter the text you want to analyze.")
print("You can enter a sentence, review, comment, or paragraph.\n")

text = input("Enter text: ")


# ============================================================
# SENTIMENT ANALYSIS
# ============================================================

blob = TextBlob(text)

polarity = blob.sentiment.polarity
subjectivity = blob.sentiment.subjectivity


# ============================================================
# DETERMINE SENTIMENT
# ============================================================

if polarity > 0.1:

    sentiment = "Positive"
    message = "The text expresses a positive opinion."

elif polarity < -0.1:

    sentiment = "Negative"
    message = "The text expresses a negative opinion."

else:

    sentiment = "Neutral"
    message = "The text is mostly neutral."


# ============================================================
# DISPLAY RESULT
# ============================================================

print("\n" + "=" * 60)

print("SENTIMENT ANALYSIS RESULT")

print("=" * 60)

print("\nInput Text:")
print(text)

print("\nSentiment:")
print(sentiment)

print(f"\nPolarity Score: {polarity:.3f}")

print(f"Subjectivity Score: {subjectivity:.3f}")

print("\nInterpretation:")
print(message)


# ============================================================
# EXPLANATION
# ============================================================

print("\n" + "=" * 60)

print("SCORE EXPLANATION")

print("=" * 60)

print("""
Polarity:
-1.0  = Very Negative
 0.0  = Neutral
+1.0  = Very Positive

Subjectivity:
0.0   = Very Objective
1.0   = Very Subjective
""")


# ============================================================
# VISUAL RESULT
# ============================================================

display(HTML(f"""

<div style="
max-width:750px;
margin:30px auto;
font-family:Arial;
">

<div style="
background:white;
padding:30px;
border-radius:20px;
border:1px solid #ddd;
">

<h2 style="text-align:center;">
Sentiment Analysis Result
</h2>

<hr>

<h3>Input Text</h3>

<div style="
background:#f5f7fb;
padding:18px;
border-radius:10px;
line-height:1.6;
">

{text}

</div>

<br>

<div style="
text-align:center;
padding:25px;
border-radius:15px;
background:#f5f7fb;
">

<h2>
{sentiment}
</h2>

<p style="font-size:18px;">
Polarity: <b>{polarity:.3f}</b>
</p>

<p style="font-size:18px;">
Subjectivity: <b>{subjectivity:.3f}</b>
</p>

</div>

<br>

<h3>Interpretation</h3>

<p>
{message}
</p>

</div>

</div>

"""))
```
